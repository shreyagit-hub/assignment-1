Reporting Pipeline

```python
def load(path):
    return pd.read_csv(path)

def clean(data):
    result = data.copy()
    result["city"] = result["city"].str.strip().str.title()
    result["attendance_percent"] = result["attendance_percent"].fillna(
        result["attendance_percent"].mean()
    )
    result["exam_date"] = pd.to_datetime(result["exam_date"], errors="coerce")
    result = result.dropna(subset=["marks"]).copy()
    result = result.drop_duplicates(keep="first").copy()
    invalid_marks = (result["marks"] < 0) | (result["marks"] > 100)
    result.loc[invalid_marks, "marks"] = np.nan
    result = result.dropna(subset=["marks"]).copy()
    result["exam_month"] = result["exam_date"].dt.to_period("M")
    result["exam_weekday"] = result["exam_date"].dt.day_name()
    result["passed"] = result["marks"] >= 40
    result["grade"] = np.where(
        result["marks"] >= 85, "A",
        np.where(result["marks"] >= 70, "B",
                 np.where(result["marks"] >= 40, "C", "F"))
    )
    result["subject"] = result["subject"].astype("category")
    result["city"] = result["city"].astype("category")
    return result

def summarise(data):
    return data.groupby("subject", observed=True)["marks"].agg(
        mean_marks="mean",
        max_marks="max",
        count="count"
    )

pipeline_summary = summarise(clean(load("../data/scores_raw.csv")))
display(pipeline_summary)

output_dir = Path("../outputs")
output_dir.mkdir(parents=True, exist_ok=True)
pipeline_summary.to_csv(output_dir / "summary_pipeline.csv")
print("Saved:", output_dir / "summary_pipeline.csv")
```

Where the cleaning could mislead

Filling in missing attendance by the column means and deleting the records with the missing and impossible marks can influence the results of the summary statistics.

Filling in the missing values of the attendance variable by the column means assumes that students with missing attendance have properties similar to those whose attendance has been captured. It is likely, that it is not correct. For example, students with missing attendance are more likely to have lower attendance rates than the rest of the sample. Then, the calculation of the means of attendance will lead to underestimation of their attendance levels. Thus, if the attendance will be later correlated with marks, the relationship between these factors will be obscured.

Deletion of the rows with the missing marks and marks below zero and above 100 protects the analysis from the influence of outliers in these variables. However, deletion of such cases implies that such deletion is random which may not hold true. Poor performers are more likely to have missing or impossible marks which leads to the increased calculated mean marks and the increased share of passes. The deletion of rows with identical observations helps to avoid the double counting of such records.
